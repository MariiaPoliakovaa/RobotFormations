```python
import numpy as np
import matplotlib.pyplot as plt

from optimization import Solver
```

```python
start_state = [0, 0, -np.pi/2]
goal_state = [0.5, 1, np.pi/2]
```

```python
solver = Solver(T=50)
```

```python
current_state = start_state

states = [current_state]
for i in range(200):
    u = solver.solve(current_state, goal_state)
    current_state = solver.dynamic_system(current_state, u, solver.dt)
    current_state = np.array(current_state).flatten()
    states.append(current_state)
```

```python
states = np.array(states)
```

```python
plt.figure(figsize=(8,8))
plt.plot(states[:, 0], states[:, 1])
```
